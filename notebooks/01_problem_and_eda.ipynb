# 01. Problem and EDA

투구 직전 정보로 `control_success` 확률을 예측합니다. 이 노트북은 데이터 계약, 시즌별 타깃 비율, 기본 파생 특성을 확인합니다. 공개 저장소에는 데이터가 포함되지 않습니다.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from src.features import add_row_local_features

DATA_PATH = PROJECT_ROOT / 'data' / 'train.csv'
train = pd.read_csv(DATA_PATH) if DATA_PATH.exists() else None
print(f'Data available: {train is not None}')
```

## Data contract

최소 열은 `season`, `control_success`, `balls_before`, `strikes_before`입니다. 선수, 주자, 이닝, 점수 차이 등 투구 직전 열을 추가할 수 있습니다.

```python
if train is not None:
    required = {'season', 'control_success', 'balls_before', 'strikes_before'}
    missing = required.difference(train.columns)
    if missing:
        raise KeyError(f'Missing required columns: {sorted(missing)}')
    display(train.groupby('season')['control_success'].agg(['count', 'mean']))
    featured = add_row_local_features(train)
    display(featured[['balls_before', 'strikes_before', 'count_total', 'count_pressure']].head())
else:
    print('Place a private train.csv in data/ to run this section.')
```
